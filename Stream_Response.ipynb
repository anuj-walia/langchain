```python
import os
from dotenv import load_dotenv
load_dotenv('conf/.env',override=True)
os.environ.get('GOOGLE_API_KEY')
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm=ChatGoogleGenerativeAI(model='gemini-1.5-flash')
prompt="Write an assay on the history of india before Mughals"
x=llm.invoke(prompt)
print(x.content)
```

```python
isitgenerator= llm.stream(prompt)
print(type(isitgenerator))
```

Chunk in below code is of type AIMessageChunk. And the above code would demonstrate that the stream api returns a generator

```python
for chunk in isitgenerator:
    print(chunk.content, sep='\n',flush=True)
```
