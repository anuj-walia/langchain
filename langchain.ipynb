The lines below enable setting user and system messages

```python
from langchain.schema import (HumanMessage, SystemMessage, AIMessage)
```

```python
messages=[
    SystemMessage(content="You are a physicist and respond in hindi"),
    HumanMessage(content="explain quantum mechanics in one sentence")
]
output=gemini_llm.invoke(messages)
print(output.content)
```

```python
from langchain.globals import set_llm_cache
from langchain_google_genai import GoogleGenerativeAI
llm=GoogleGenerativeAI(model='gemini-1.5-flash')
```

The lines below will help setting caching for the prompts and the outputs

```python
%%time
from langchain.cache import InMemoryCache
set_llm_cache(InMemoryCache())
prompt='How to lear apache flink step by step with reference materials including video tutorials'
llm.invoke(prompt)
```
