```python
import os
from dotenv import load_dotenv
from langsmith.schemas import Prompt

load_dotenv('conf/.env',override=True)
os.environ.get('GOOGLE_API_KEY')
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
from langchain.prompts import PromptTemplate, ChatPromptTemplate, HumanMessagePromptTemplate,SystemMessagePromptTemplate
from langchain_core.messages import SystemMessage
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
template ='''
You are an experienced virologist
Write a few sentences aboout the following virus "{virus}" in {language}
'''
prompt_template = PromptTemplate.from_template(template=template)
prompt = prompt_template.format(virus='hiv',language='german')
llm=ChatGoogleGenerativeAI(model='gemini-1.5-flash',temperature=0)
output=llm.invoke(prompt)
print(output.content)
```

```python
chat_template = ChatPromptTemplate.from_messages([
    SystemMessage(content='You respond only in json format'),
    HumanMessagePromptTemplate.from_template('Top {n} countries in {area} by population.')
])
```

```python
messages=chat_template.format(n='10', area='Europe')            
print(messages)
```

```python
messages=chat_template.format_messages(n='10', area='Europe')
print(messages)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model='gemini-1.5-flash')
output=llm.invoke(messages)
print(output.content)
```
